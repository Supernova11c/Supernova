````python
import torch
import torch.nn as nn
import numpy as np
import math
import random

SEED = 411
torch.manual_seed(SEED)
np.random.seed(SEED)
random.seed(SEED)

DTYPE = torch.float64
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Photon-1 environment ready")
print("Device:", DEVICE)
print("Precision:", DTYPE)
````

````python
class Photon1MoE(nn.Module):

    def __init__(self):
        super().__init__()

        # Exactly one trainable parameter.
        self.log_gain = nn.Parameter(
            torch.tensor(0.15, dtype=DTYPE)
        )

        self.experts = {
            "E": self.expert_energy,
            "M": self.expert_mass,
            "C": self.expert_speed,
            "C2": self.expert_speed_squared
        }

    def calibration(self):
        return torch.exp(self.log_gain)

    def expert_energy(self, m, c):
        return m * (c ** 2)

    def expert_mass(self, E, c):
        return E / (c ** 2)

    def expert_speed(self, E, m):
        return torch.sqrt(E / m)

    def expert_speed_squared(self, E, m):
        return E / m

    def forward(self, a, b, target):

        target = target.upper().replace("²", "2")

        if target not in self.experts:
            raise ValueError(
                "Target must be E, M, C, or C2"
            )

        a = torch.as_tensor(
            a, dtype=DTYPE, device=DEVICE
        )

        b = torch.as_tensor(
            b, dtype=DTYPE, device=DEVICE
        )

        if torch.any(a <= 0) or torch.any(b <= 0):
            raise ValueError(
                "Inputs must be positive."
            )

        # MoE router selects one expert.
        expert = self.experts[target]

        # Execute selected mathematical expert.
        raw_result = expert(a, b)

        # Shared one-parameter calibration.
        return raw_result * self.calibration()


photon = Photon1MoE().to(DEVICE)

trainable_count = sum(
    p.numel()
    for p in photon.parameters()
    if p.requires_grad
)

print("Photon-1 MoE created")
print("Experts:", list(photon.experts.keys()))
print("Trainable parameters:", trainable_count)

assert trainable_count == 1
````

````python
N_TRAIN = 12000

# Random positive mass values.
train_m = torch.pow(
    10.0,
    torch.empty(N_TRAIN, dtype=DTYPE).uniform_(-6, 6)
).to(DEVICE)

# Random positive speed values for algebraic testing.
train_c = torch.pow(
    10.0,
    torch.empty(N_TRAIN, dtype=DTYPE).uniform_(3, 9)
).to(DEVICE)

# Generate energy using Einstein's equation.
train_E = train_m * train_c.square()

# Four expert training targets.
training_examples = {
    "E": (train_m, train_c, train_E),
    "M": (train_E, train_c, train_m),
    "C": (train_E, train_m, train_c),
    "C2": (train_E, train_m, train_c.square())
}

print("Training examples generated")
print("Examples per expert:", N_TRAIN)
print("Total expert examples:", N_TRAIN * 4)
````

````python
optimizer = torch.optim.Adam(
    photon.parameters(),
    lr=0.03
)

EPOCHS = 300

history = []

for epoch in range(EPOCHS):

    optimizer.zero_grad()

    losses = []

    for target, (a, b, expected) in training_examples.items():

        predicted = photon(a, b, target)

        loss = torch.mean(
            (torch.log(predicted) - torch.log(expected)) ** 2
        )

        losses.append(loss)

    total_loss = torch.stack(losses).mean()

    total_loss.backward()
    optimizer.step()

    history.append(total_loss.item())

    if (epoch + 1) % 50 == 0:
        print(
            f"Epoch {epoch + 1:3d}/{EPOCHS} | "
            f"Loss: {total_loss.item():.12e} | "
            f"Calibration: {photon.calibration().item():.12f}"
        )

print("\nTraining completed.")
print("Final calibration:", photon.calibration().item())
print(
    "Trainable parameters:",
    sum(p.numel() for p in photon.parameters() if p.requires_grad)
)
````

````python
import torch
import torch.nn as nn
import numpy as np
import random
import math

SEED = 411
torch.manual_seed(SEED)
np.random.seed(SEED)
random.seed(SEED)

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

DTYPE = torch.float64

print("Supernova TeraiLLM Photon-1 V2")
print("Device:", DEVICE)
print("Precision:", DTYPE)
````

````python
class Photon1V2(nn.Module):

    def __init__(self):
        super().__init__()

        # Exactly one trainable scalar.
        self.log_gain = nn.Parameter(
            torch.tensor(math.log(1.05), dtype=DTYPE)
        )

    def gain(self):
        return torch.exp(self.log_gain)

    def forward(self, a, b, target):

        target = target.upper().replace("²", "2")

        if target not in ("E", "M", "C", "C2"):
            raise ValueError("Target must be E, M, C, or C2.")

        a = torch.as_tensor(a, dtype=DTYPE, device=DEVICE)
        b = torch.as_tensor(b, dtype=DTYPE, device=DEVICE)

        if not torch.all(torch.isfinite(a)):
            raise ValueError("First input must be finite.")

        if not torch.all(torch.isfinite(b)):
            raise ValueError("Second input must be finite.")

        if torch.any(a <= 0) or torch.any(b <= 0):
            raise ValueError("Both inputs must be positive.")

        if target == "E":
            # a = mass (kg), b = speed (m/s)
            raw = a * b.square()

        elif target == "M":
            # a = energy (J), b = speed (m/s)
            raw = a / b.square()

        elif target == "C":
            # a = energy (J), b = mass (kg)
            raw = torch.sqrt(a / b)

        else:
            # a = energy (J), b = mass (kg)
            raw = a / b

        result = raw * self.gain()

        if not torch.all(torch.isfinite(result)):
            raise ValueError("Result exceeds the supported numerical range.")

        return result


photon_v2 = Photon1V2().to(DEVICE)

parameter_count = sum(
    p.numel()
    for p in photon_v2.parameters()
    if p.requires_grad
)

print("Experts: E, M, C, C2")
print("Trainable parameters:", parameter_count)

assert parameter_count == 1
````

````python
N_TRAIN = 12000

train_m = torch.pow(
    10.0,
    torch.empty(N_TRAIN, dtype=DTYPE).uniform_(-6, 6)
).to(DEVICE)

train_c = torch.pow(
    10.0,
    torch.empty(N_TRAIN, dtype=DTYPE).uniform_(3, 9)
).to(DEVICE)

train_E = train_m * train_c.square()
train_C2 = train_c.square()

training_data = {
    "E": (train_m, train_c, train_E),
    "M": (train_E, train_c, train_m),
    "C": (train_E, train_m, train_c),
    "C2": (train_E, train_m, train_C2)
}

print("Training data ready.")
print("Samples per expert:", N_TRAIN)
print("Total samples:", N_TRAIN * 4)
````

````python
optimizer = torch.optim.Adam(
    photon_v2.parameters(),
    lr=0.03
)

EPOCHS = 300
history = []

for epoch in range(EPOCHS):

    optimizer.zero_grad()

    losses = []

    for target, (a, b, expected) in training_data.items():

        predicted = photon_v2(a, b, target)

        loss = torch.mean(
            (torch.log(predicted) - torch.log(expected)) ** 2
        )

        losses.append(loss)

    total_loss = torch.stack(losses).mean()

    total_loss.backward()
    optimizer.step()

    history.append(total_loss.item())

    if (epoch + 1) % 50 == 0:

        print(
            f"Epoch {epoch + 1}/{EPOCHS} | "
            f"Loss: {total_loss.item():.12e} | "
            f"Gain: {photon_v2.gain().item():.12f}"
        )

print("\nTraining completed.")
print("Learned gain:", photon_v2.gain().item())
print("Trainable parameters:", parameter_count)
````

````python
N_TEST = 10000

rng = np.random.default_rng(20261101)

test_m = 10.0 ** rng.uniform(-6, 6, N_TEST)
test_c = 10.0 ** rng.uniform(3, 9, N_TEST)

test_E = test_m * test_c**2
test_C2 = test_c**2

test_data = {
    "E": (test_m, test_c, test_E),
    "M": (test_E, test_c, test_m),
    "C": (test_E, test_m, test_c),
    "C2": (test_E, test_m, test_C2)
}


def benchmark(model, data):

    report = {}

    model.eval()

    with torch.no_grad():

        for target, (a, b, expected) in data.items():

            prediction = model(a, b, target).cpu().numpy()

            expected = np.asarray(expected)

            relative_error = (
                np.abs(prediction - expected)
                / np.abs(expected)
            ) * 100.0

            report[target] = {
                "mean": float(np.mean(relative_error)),
                "median": float(np.median(relative_error)),
                "maximum": float(np.max(relative_error)),
                "within_0_01": int(np.sum(relative_error <= 0.01)),
                "within_0_1": int(np.sum(relative_error <= 0.1)),
                "within_1": int(np.sum(relative_error <= 1.0)),
                "above_1": int(np.sum(relative_error > 1.0))
            }

    return report


baseline_report = benchmark(baseline, test_data)
trained_report = benchmark(photon_v2, test_data)

print("=" * 85)
print("BASELINE VS TRAINED PHOTON-1 V2")
print("=" * 85)

for target in ("E", "M", "C", "C2"):

    print(f"\nEXPERT: {target}")

    for label, report in (
        ("Baseline", baseline_report),
        ("Trained", trained_report)
    ):

        r = report[target]

        print(
            f"{label:10s} | "
            f"Mean: {r['mean']:.12e}% | "
            f"Max: {r['maximum']:.12e}% | "
            f"Within 1%: {r['within_1']}/{N_TEST}"
        )
````

````python
import math
import numpy as np
import torch
import torch.nn as nn

SEED = 411
np.random.seed(SEED)
torch.manual_seed(SEED)

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

DTYPE = torch.float64

C_LIGHT = 299792458.0  # m/s

print("Supernova TeraiLLM Photon-1 V3")
print("Device:", DEVICE)
print("Precision:", DTYPE)
print("Speed of light:", C_LIGHT, "m/s")
````

````python
UNIT_TABLE = {
    "kg": ("mass", 1.0),
    "g": ("mass", 0.001),

    "j": ("energy", 1.0),
    "kj": ("energy", 1000.0),

    "m/s": ("speed", 1.0),
    "km/s": ("speed", 1000.0),

    "m2/s2": ("speed_squared", 1.0),
    "m²/s²": ("speed_squared", 1.0),
    "km2/s2": ("speed_squared", 1_000_000.0),
    "km²/s²": ("speed_squared", 1_000_000.0)
}


def normalize_unit(unit):
    return unit.strip().lower().replace(" ", "")


def convert_to_si(value, unit, expected_dimension):

    unit = normalize_unit(unit)

    if unit not in UNIT_TABLE:
        raise ValueError(f"Unsupported unit: {unit}")

    dimension, factor = UNIT_TABLE[unit]

    if dimension != expected_dimension:
        raise ValueError(
            f"Unit {unit} is incompatible with "
            f"{expected_dimension}."
        )

    value = float(value)

    if not math.isfinite(value) or value <= 0:
        raise ValueError("Input must be finite and positive.")

    result = value * factor

    if not math.isfinite(result):
        raise ValueError("Unit conversion overflow.")

    return result


print("Unit conversion system ready.")
print("10 g =", convert_to_si(10, "g", "mass"), "kg")
print("2 km/s =", convert_to_si(2, "km/s", "speed"), "m/s")
````

````python
class PhotonExperts:

    @staticmethod
    def energy(mass, speed):
        return mass * speed**2

    @staticmethod
    def mass(energy, speed):
        return energy / speed**2

    @staticmethod
    def speed(energy, mass):
        return math.sqrt(energy / mass)

    @staticmethod
    def speed_squared(energy, mass):
        return energy / mass


EXPERTS = {
    "E": PhotonExperts.energy,
    "M": PhotonExperts.mass,
    "C": PhotonExperts.speed,
    "C2": PhotonExperts.speed_squared
}

print("Photon experts:", list(EXPERTS.keys()))
````

````python
class PhotonCoreV3:

    def __init__(self):
        self.experts = EXPERTS

    def route(self, target):

        target = target.upper().strip()
        target = target.replace("²", "2")

        if target not in self.experts:
            raise ValueError(
                "Unknown target. Choose E, M, C, or C2."
            )

        return target

    def solve(self, target, value1, unit1, value2, unit2):

        target = self.route(target)

        if target == "E":

            a = convert_to_si(value1, unit1, "mass")
            b = convert_to_si(value2, unit2, "speed")

            formula = "E = mc²"
            result = self.experts["E"](a, b)

            steps = [
                f"E = m × c²",
                f"E = {a} × {b}²",
                f"E = {result} J"
            ]

            output_unit = "J"

        elif target == "M":

            a = convert_to_si(value1, unit1, "energy")
            b = convert_to_si(value2, unit2, "speed")

            formula = "m = E/c²"
            result = self.experts["M"](a, b)

            steps = [
                "m = E / c²",
                f"m = {a} / {b}²",
                f"m = {result} kg"
            ]

            output_unit = "kg"

        elif target == "C":

            a = convert_to_si(value1, unit1, "energy")
            b = convert_to_si(value2, unit2, "mass")

            formula = "c = √(E/m)"
            result = self.experts["C"](a, b)

            steps = [
                "c = √(E/m)",
                f"c = √({a}/{b})",
                f"c = {result} m/s"
            ]

            output_unit = "m/s"

        else:

            a = convert_to_si(value1, unit1, "energy")
            b = convert_to_si(value2, unit2, "mass")

            formula = "c² = E/m"
            result = self.experts["C2"](a, b)

            steps = [
                "c² = E/m",
                f"c² = {a}/{b}",
                f"c² = {result} m²/s²"
            ]

            output_unit = "m²/s²"

        if not math.isfinite(result) or result <= 0:
            raise ValueError(
                "Result is outside the supported numerical range."
            )

        return {
            "expert": target,
            "formula": formula,
            "result": result,
            "unit": output_unit,
            "steps": steps
        }


photon_core = PhotonCoreV3()

print("Photon Core V3 initialized.")
print("Available experts:", list(photon_core.experts.keys()))
````

````python
UNIT_TABLE = {
    # Mass
    "kg": ("mass", 1.0),
    "g": ("mass", 0.001),

    # Distance
    "m": ("distance", 1.0),
    "cm": ("distance", 0.01),
    "km": ("distance", 1000.0),

    # Time
    "s": ("time", 1.0),
    "min": ("time", 60.0),
    "h": ("time", 3600.0),

    # Speed
    "m/s": ("speed", 1.0),
    "km/s": ("speed", 1000.0),
    "km/h": ("speed", 1000 / 3600),

    # Acceleration
    "m/s2": ("acceleration", 1.0),
    "m/s^2": ("acceleration", 1.0),
    "cm/s2": ("acceleration", 0.01),

    # Volume
    "m3": ("volume", 1.0),
    "m^3": ("volume", 1.0),
    "l": ("volume", 0.001),
    "cm3": ("volume", 1e-6),

    # Density
    "kg/m3": ("density", 1.0),
    "kg/m^3": ("density", 1.0),
    "g/cm3": ("density", 1000.0),

    # Force
    "n": ("force", 1.0),
    "kn": ("force", 1000.0),

    # Energy and work
    "j": ("energy", 1.0),
    "kj": ("energy", 1000.0),

    # Power
    "w": ("power", 1.0),
    "kw": ("power", 1000.0)
}


VARIABLE_UNITS = {
    "mass": ("mass", "kg"),
    "energy": ("energy", "J"),
    "speed": ("speed", "m/s"),
    "distance": ("distance", "m"),
    "time": ("time", "s"),
    "density": ("density", "kg/m³"),
    "volume": ("volume", "m³"),
    "force": ("force", "N"),
    "acceleration": ("acceleration", "m/s²"),
    "work": ("energy", "J"),
    "power": ("power", "W")
}


def to_si(value, unit, expected_dimension):

    unit = unit.strip().lower().replace(" ", "")

    if unit not in UNIT_TABLE:
        raise ValueError(f"Unsupported unit: {unit}")

    dimension, factor = UNIT_TABLE[unit]

    if dimension != expected_dimension:
        raise ValueError(
            f"Unit '{unit}' is incompatible with "
            f"'{expected_dimension}'."
        )

    value = float(value)

    if not math.isfinite(value):
        raise ValueError("Input must be finite.")

    return value * factor


def from_si(value, variable):

    dimension, unit = VARIABLE_UNITS[variable]

    factor = next(
        factor
        for d, factor in UNIT_TABLE.values()
        if False
    ) if False else None

    # SI base units have conversion factor 1.
    return value, unit


print("Unit system initialized.")
print("1 km/h =", to_si(1, "km/h", "speed"), "m/s")
print("1 g/cm3 =", to_si(1, "g/cm3", "density"), "kg/m³")
print("1 kJ =", to_si(1, "kJ", "energy"), "J")
````

````python
def safe_divide(a, b):
    if b == 0:
        raise ValueError("Division by zero is not allowed.")
    return a / b


EQUATIONS = {

    "einstein": {
        "variables": {"energy", "mass", "speed"},
        "solvers": {
            "energy": lambda x: x["mass"] * x["speed"]**2,
            "mass": lambda x: safe_divide(
                x["energy"], x["speed"]**2
            ),
            "speed": lambda x: math.sqrt(
                safe_divide(x["energy"], x["mass"])
            )
        }
    },

    "speed": {
        "variables": {"speed", "distance", "time"},
        "solvers": {
            "speed": lambda x: safe_divide(
                x["distance"], x["time"]
            ),
            "distance": lambda x: x["speed"] * x["time"],
            "time": lambda x: safe_divide(
                x["distance"], x["speed"]
            )
        }
    },

    "density": {
        "variables": {"density", "mass", "volume"},
        "solvers": {
            "density": lambda x: safe_divide(
                x["mass"], x["volume"]
            ),
            "mass": lambda x: x["density"] * x["volume"],
            "volume": lambda x: safe_divide(
                x["mass"], x["density"]
            )
        }
    },

    "newton": {
        "variables": {"force", "mass", "acceleration"},
        "solvers": {
            "force": lambda x: x["mass"] * x["acceleration"],
            "mass": lambda x: safe_divide(
                x["force"], x["acceleration"]
            ),
            "acceleration": lambda x: safe_divide(
                x["force"], x["mass"]
            )
        }
    },

    "work": {
        "variables": {"work", "force", "distance"},
        "solvers": {
            "work": lambda x: x["force"] * x["distance"],
            "force": lambda x: safe_divide(
                x["work"], x["distance"]
            ),
            "distance": lambda x: safe_divide(
                x["work"], x["force"]
            )
        }
    },

    "power": {
        "variables": {"power", "work", "time"},
        "solvers": {
            "power": lambda x: safe_divide(
                x["work"], x["time"]
            ),
            "work": lambda x: x["power"] * x["time"],
            "time": lambda x: safe_divide(
                x["work"], x["power"]
            )
        }
    }
}


print("Equation library loaded.")

for name, equation in EQUATIONS.items():
    print(name, ":", sorted(equation["variables"]))
````

````python
ALIASES = {
    "e": "energy",
    "energy": "energy",

    "m": "mass",
    "mass": "mass",

    "c": "speed",
    "v": "speed",
    "speed": "speed",

    "d": "distance",
    "distance": "distance",

    "t": "time",
    "time": "time",

    "rho": "density",
    "ρ": "density",
    "density": "density",

    "v_volume": "volume",
    "volume": "volume",

    "f": "force",
    "force": "force",

    "a": "acceleration",
    "acceleration": "acceleration",

    "w_work": "work",
    "work": "work",

    "p": "power",
    "power": "power"
}


def normalize_variable(variable):

    key = variable.strip().lower()

    if key not in ALIASES:
        raise ValueError(
            f"Unknown variable: {variable}"
        )

    return ALIASES[key]


print("Variable recognition system ready.")
````

````python
def select_equation(target, known_variables, equation_name=None):

    target = normalize_variable(target)

    known_variables = {
        normalize_variable(v)
        for v in known_variables
    }

    candidates = []

    for name, equation in EQUATIONS.items():

        variables = equation["variables"]

        if target not in equation["solvers"]:
            continue

        required = variables - {target}

        if required == known_variables:
            candidates.append(name)

    if equation_name is not None:

        equation_name = equation_name.lower().strip()

        if equation_name not in candidates:
            raise ValueError(
                f"{equation_name} cannot solve {target} "
                f"from {sorted(known_variables)}."
            )

        return equation_name

    if len(candidates) == 0:
        raise ValueError(
            "No compatible equation found for these variables."
        )

    if len(candidates) > 1:
        raise ValueError(
            "Ambiguous equation. Candidates: "
            + ", ".join(candidates)
            + ". Specify the equation name."
        )

    return candidates[0]


# Demonstrations
print(
    "Energy:",
    select_equation("energy", ["mass", "speed"])
)

print(
    "Acceleration:",
    select_equation("a", ["force", "mass"])
)

print(
    "Density:",
    select_equation("rho", ["mass", "volume"])
)
````

````python
class PhotonCoreV4:

    def __init__(self):
        self.equations = EQUATIONS

    def solve(
        self,
        target,
        known_values,
        equation_name=None
    ):

        target = normalize_variable(target)

        normalized = {}

        for variable, data in known_values.items():

            variable = normalize_variable(variable)

            if variable == target:
                raise ValueError(
                    "The target cannot also be a known variable."
                )

            if variable in normalized:
                raise ValueError(
                    f"Duplicate variable: {variable}"
                )

            value, unit = data

            dimension, _ = VARIABLE_UNITS[variable]

            normalized[variable] = to_si(
                value, unit, dimension
            )

        selected = select_equation(
            target,
            normalized.keys(),
            equation_name
        )

        solver = self.equations[selected]["solvers"][target]

        result = solver(normalized)

        if not math.isfinite(result):
            raise ValueError(
                "Result is outside the supported numerical range."
            )

        if target in {
            "mass", "energy", "time", "volume", "density"
        } and result <= 0:
            raise ValueError(
                f"Calculated {target} must be positive."
            )

        if target in {"speed", "distance"} and result < 0:
            raise ValueError(
                f"Calculated {target} cannot be negative."
            )

        output_unit = VARIABLE_UNITS[target][1]

        return {
            "equation": selected,
            "target": target,
            "known_values_si": normalized,
            "result": result,
            "unit": output_unit
        }


photon_v4 = PhotonCoreV4()

print("Photon Core V4 initialized.")
print("Available equations:", list(EQUATIONS.keys()))
````

````python
# ============================================================
# PHOTON-1 V5 — UNIT SYSTEM
# All internal calculations use SI units.
# ============================================================

ALIASES = {
    "e": "energy",
    "energy": "energy",
    "m": "mass",
    "mass": "mass",
    "c": "speed",
    "v": "speed",
    "speed": "speed",
    "d": "distance",
    "distance": "distance",
    "t": "time",
    "time": "time",
    "rho": "density",
    "ρ": "density",
    "density": "density",
    "f": "force",
    "force": "force",
    "a": "acceleration",
    "acceleration": "acceleration",
    "w": "work",
    "work": "work",
    "p": "power",
    "power": "power",
}

# Conversion factors: supplied value * factor = SI value
UNIT_TABLE = {
    "energy": {
        "J": 1.0,
        "kJ": 1000.0,
    },
    "mass": {
        "kg": 1.0,
        "g": 0.001,
    },
    "speed": {
        "m/s": 1.0,
        "km/s": 1000.0,
        "km/h": 1 / 3.6,
    },
    "distance": {
        "m": 1.0,
        "cm": 0.01,
        "km": 1000.0,
    },
    "time": {
        "s": 1.0,
        "min": 60.0,
        "h": 3600.0,
    },
    "density": {
        "kg/m3": 1.0,
        "g/cm3": 1000.0,
    },
    "volume": {
        "m3": 1.0,
        "L": 0.001,
        "cm3": 1e-6,
    },
    "force": {
        "N": 1.0,
        "kN": 1000.0,
    },
    "acceleration": {
        "m/s2": 1.0,
        "cm/s2": 0.01,
    },
    "work": {
        "J": 1.0,
        "kJ": 1000.0,
    },
    "power": {
        "W": 1.0,
        "kW": 1000.0,
    },
}

def normalize_variable(name):
    key = str(name).strip().lower()
    if key not in ALIASES:
        raise ValueError(f"Unknown variable: {name}")
    return ALIASES[key]


def to_si(variable, value, unit):
    variable = normalize_variable(variable)

    if variable not in UNIT_TABLE:
        raise ValueError(f"No unit definition for {variable}")

    if unit not in UNIT_TABLE[variable]:
        raise ValueError(
            f"Invalid unit {unit!r} for variable {variable!r}"
        )

    value = float(value)

    if not math.isfinite(value):
        raise ValueError("Input must be finite")

    return value * UNIT_TABLE[variable][unit]


def from_si(variable, value, unit):
    variable = normalize_variable(variable)

    if unit not in UNIT_TABLE[variable]:
        raise ValueError(
            f"Invalid output unit {unit!r} for {variable!r}"
        )

    return value / UNIT_TABLE[variable][unit]


def safe_divide(a, b):
    if b == 0:
        raise ValueError("Division by zero is not allowed")
    return a / b


# ============================================================
# EQUATION REGISTRY
# ============================================================

EQUATIONS = {
    "einstein": {
        "variables": {"energy", "mass", "speed"},
        "solvers": {
            "energy": lambda x: x["mass"] * x["speed"]**2,
            "mass": lambda x: safe_divide(
                x["energy"], x["speed"]**2
            ),
            "speed": lambda x: math.sqrt(
                safe_divide(x["energy"], x["mass"])
            ),
        },
    },

    "speed": {
        "variables": {"speed", "distance", "time"},
        "solvers": {
            "speed": lambda x: safe_divide(
                x["distance"], x["time"]
            ),
            "distance": lambda x: x["speed"] * x["time"],
            "time": lambda x: safe_divide(
                x["distance"], x["speed"]
            ),
        },
    },

    "density": {
        "variables": {"density", "mass", "volume"},
        "solvers": {
            "density": lambda x: safe_divide(
                x["mass"], x["volume"]
            ),
            "mass": lambda x: x["density"] * x["volume"],
            "volume": lambda x: safe_divide(
                x["mass"], x["density"]
            ),
        },
    },

    "newton": {
        "variables": {"force", "mass", "acceleration"},
        "solvers": {
            "force": lambda x: x["mass"] * x["acceleration"],
            "mass": lambda x: safe_divide(
                x["force"], x["acceleration"]
            ),
            "acceleration": lambda x: safe_divide(
                x["force"], x["mass"]
            ),
        },
    },

    "work": {
        "variables": {"work", "force", "distance"},
        "solvers": {
            "work": lambda x: x["force"] * x["distance"],
            "force": lambda x: safe_divide(
                x["work"], x["distance"]
            ),
            "distance": lambda x: safe_divide(
                x["work"], x["force"]
            ),
        },
    },

    "power": {
        "variables": {"power", "work", "time"},
        "solvers": {
            "power": lambda x: safe_divide(
                x["work"], x["time"]
            ),
            "work": lambda x: x["power"] * x["time"],
            "time": lambda x: safe_divide(
                x["work"], x["power"]
            ),
        },
    },
}

print("Six equations registered.")
print("Total solving directions:",
      sum(len(eq["solvers"]) for eq in EQUATIONS.values()))
````

````python
# ============================================================
# PHOTON-1 V5 — ONE-PARAMETER RESEARCH BRANCH
# ============================================================

class AdaptiveTolerance:

    def __init__(self):
        # Exactly ONE trainable parameter.
        self.log_tolerance = float(np.log(1e-6))

        # Fixed hyperparameter, NOT trainable.
        self.temperature = 0.5

    @property
    def tolerance(self):
        return float(np.exp(self.log_tolerance))

    def probability_valid(self, residual):
        residual = max(float(residual), 1e-16)

        log_error = np.log(residual)
        z = (self.log_tolerance - log_error) / self.temperature

        z = np.clip(z, -60, 60)
        return float(1.0 / (1.0 + np.exp(-z)))

    def train(self, residuals, labels, epochs=2000, lr=0.03):
        residuals = np.asarray(residuals, dtype=np.float64)
        labels = np.asarray(labels, dtype=np.float64)

        if len(residuals) != len(labels):
            raise ValueError("Residual and label lengths differ")

        if not np.all(np.isfinite(residuals)):
            raise ValueError("Training residuals must be finite")

        if not np.all((labels == 0) | (labels == 1)):
            raise ValueError("Labels must be 0 or 1")

        log_errors = np.log(np.maximum(residuals, 1e-16))

        history = []

        for epoch in range(epochs):
            z = (
                self.log_tolerance - log_errors
            ) / self.temperature

            z = np.clip(z, -60, 60)
            probabilities = 1.0 / (1.0 + np.exp(-z))

            # Binary cross-entropy.
            eps = 1e-12
            loss = -np.mean(
                labels * np.log(probabilities + eps)
                + (1 - labels) * np.log(1 - probabilities + eps)
            )

            # Analytic gradient for the single parameter.
            gradient = np.mean(
                probabilities - labels
            ) / self.temperature

            self.log_tolerance -= lr * gradient

            if not np.isfinite(self.log_tolerance):
                raise RuntimeError("Training became numerically unstable")

            if epoch % 200 == 0:
                history.append(float(loss))

        return history


# Generate labelled research examples.
# Clean examples have tiny residuals.
# Corrupted examples have larger residuals.

rng = np.random.default_rng(42)

clean_residuals = 10 ** rng.uniform(-16, -14, 5000)
corrupted_residuals = 10 ** rng.uniform(-8, -2, 5000)

training_residuals = np.concatenate([
    clean_residuals,
    corrupted_residuals
])

training_labels = np.concatenate([
    np.ones(len(clean_residuals)),
    np.zeros(len(corrupted_residuals))
])

# Shuffle deterministically.
indices = rng.permutation(len(training_residuals))
training_residuals = training_residuals[indices]
training_labels = training_labels[indices]

# Independent held-out test data.
test_clean = 10 ** rng.uniform(-16, -14, 1000)
test_corrupted = 10 ** rng.uniform(-8, -2, 1000)

test_residuals = np.concatenate([test_clean, test_corrupted])
test_labels = np.concatenate([
    np.ones(len(test_clean)),
    np.zeros(len(test_corrupted))
])

model = AdaptiveTolerance()
training_history = model.train(
    training_residuals,
    training_labels
)

predictions = np.array([
    model.probability_valid(r) >= 0.5
    for r in test_residuals
])

accuracy = np.mean(predictions == test_labels) * 100

print("=" * 60)
print("PHOTON-1 V5 ADAPTIVE TOLERANCE TRAINING")
print("=" * 60)
print("Trainable parameters:", 1)
print("Learned log_tolerance:", model.log_tolerance)
print("Learned tolerance:", model.tolerance)
print("Held-out classification accuracy:", accuracy, "%")
print("Training loss checkpoints:", training_history)
````

````python
# ============================================================
# SUPERNOVA TERAILLM PHOTON-1 V5
# EXACT MATHEMATICAL CORE
# ============================================================

def select_equation(target, known_variables, equation_name=None):

    target = normalize_variable(target)

    known_variables = {
        normalize_variable(v)
        for v in known_variables
    }

    if equation_name is not None:

        if equation_name not in EQUATIONS:
            raise ValueError("Unknown equation")

        equation = EQUATIONS[equation_name]

        if target not in equation["solvers"]:
            raise ValueError(
                "Target is not supported by this equation"
            )

        expected = equation["variables"] - {target}

        if known_variables != expected:
            raise ValueError(
                "Incompatible equation or missing variables"
            )

        return equation_name

    matches = []

    for name, equation in EQUATIONS.items():

        expected = equation["variables"] - {target}

        if (
            target in equation["solvers"]
            and known_variables == expected
        ):
            matches.append(name)

    if len(matches) == 0:
        raise ValueError("No compatible equation found")

    if len(matches) > 1:
        raise ValueError(
            "Ambiguous equation. Specify equation_name explicitly."
        )

    return matches[0]


def equation_residual(equation_name, values):

    if equation_name == "einstein":
        lhs = values["energy"]
        rhs = values["mass"] * values["speed"]**2

    elif equation_name == "speed":
        lhs = values["speed"]
        rhs = values["distance"] / values["time"]

    elif equation_name == "density":
        lhs = values["density"]
        rhs = values["mass"] / values["volume"]

    elif equation_name == "newton":
        lhs = values["force"]
        rhs = values["mass"] * values["acceleration"]

    elif equation_name == "work":
        lhs = values["work"]
        rhs = values["force"] * values["distance"]

    elif equation_name == "power":
        lhs = values["power"]
        rhs = values["work"] / values["time"]

    else:
        raise ValueError("Unknown equation")

    scale = max(abs(lhs), abs(rhs), 1e-300)

    return abs(lhs - rhs) / scale


class PhotonCoreV5:

    def __init__(self, adaptive_model):
        self.adaptive_model = adaptive_model

    def validate_result(self, equation_name, values):

        residual = equation_residual(
            equation_name,
            values
        )

        probability = self.adaptive_model.probability_valid(
            residual
        )

        accepted = probability >= 0.5

        return {
            "accepted": accepted,
            "residual": residual,
            "tolerance": self.adaptive_model.tolerance,
            "probability": probability
        }

    def solve(
        self,
        target,
        known_values,
        equation_name=None,
        target_unit=None
    ):

        target = normalize_variable(target)

        if not isinstance(known_values, dict):
            raise ValueError(
                "Known values must be a dictionary"
            )

        if not known_values:
            raise ValueError(
                "Known values cannot be empty"
            )

        # Normalize variables and convert inputs to SI.
        values = {}

        for name, item in known_values.items():

            variable = normalize_variable(name)

            if variable in values:
                raise ValueError(
                    "Duplicate variable supplied"
                )

            if (
                not isinstance(item, (tuple, list))
                or len(item) != 2
            ):
                raise ValueError(
                    f"{variable} must be supplied as (value, unit)"
                )

            value, unit = item

            values[variable] = to_si(
                variable,
                value,
                unit
            )

        if target in values:
            raise ValueError(
                "Target variable is already provided"
            )

        # Automatic equation selection.
        selected = select_equation(
            target,
            set(values.keys()),
            equation_name
        )

        equation = EQUATIONS[selected]

        nonnegative = {
            "energy",
            "mass",
            "speed",
            "distance",
            "time",
            "density",
            "volume",
            "force",
            "work",
            "power"
        }

        # Validate known values.
        for variable, value in values.items():

            if variable in nonnegative and value < 0:
                raise ValueError(
                    f"{variable} cannot be negative"
                )

        # Exact mathematical calculation.
        result_si = equation["solvers"][target](values)

        if not math.isfinite(result_si):
            raise ValueError(
                "Calculation produced a non-finite result"
            )

        if target in nonnegative and result_si < 0:
            raise ValueError(
                "Calculated result cannot be negative"
            )

        # Reconstruct complete equation values.
        complete_values = dict(values)
        complete_values[target] = result_si

        # Adaptive numerical validation.
        validation = self.validate_result(
            selected,
            complete_values
        )

        if not validation["accepted"]:
            raise ArithmeticError(
                "Result failed adaptive numerical validation"
            )

        # Default output units.
        default_units = {
            "energy": "J",
            "mass": "kg",
            "speed": "m/s",
            "distance": "m",
            "time": "s",
            "density": "kg/m3",
            "volume": "m3",
            "force": "N",
            "acceleration": "m/s2",
            "work": "J",
            "power": "W"
        }

        output_unit = (
            target_unit
            if target_unit is not None
            else default_units[target]
        )

        result = from_si(
            target,
            result_si,
            output_unit
        )

        formulas = {
            "einstein": "E = mc²",
            "speed": "v = d/t",
            "density": "ρ = m/V",
            "newton": "F = ma",
            "work": "W = Fd",
            "power": "P = W/t"
        }

        return {
            "equation": selected,
            "target": target,
            "formula": formulas[selected],
            "result": result,
            "unit": output_unit,
            "si_result": result_si,
            "residual": validation["residual"],
            "adaptive_tolerance": validation["tolerance"],
            "validation_probability": validation["probability"]
        }


# Initialize Photon-1 V5.
photon = PhotonCoreV5(model)

print("=" * 60)
print("PHOTON-1 V5 MATHEMATICAL CORE")
print("=" * 60)

print("Equations:", len(EQUATIONS))

print(
    "Solving directions:",
    sum(
        len(eq["solvers"])
        for eq in EQUATIONS.values()
    )
)

print("Trainable research parameters:", 1)
print("Exact core uses learned arithmetic:", False)
print("Volume variable registration:", "Enabled")
print("Core initialization: SUCCESS")
````

````python
# ============================================================
# SUPERNOVA TERAILLM PHOTON-1 V5
# CELL 9: ADAPTIVE vs FIXED TOLERANCE
# ============================================================

import numpy as np

rng = np.random.default_rng(2026)

print("=" * 75)
print("PHOTON-1 V5 — ADAPTIVE vs FIXED TOLERANCE")
print("=" * 75)

# ------------------------------------------------------------
# 1. Locate the trained AdaptiveTolerance instance
# ------------------------------------------------------------

adaptive_model = None

for name, obj in list(globals().items()):
    if isinstance(obj, AdaptiveTolerance):
        adaptive_model = obj
        print("Adaptive model found:", name)
        break

if adaptive_model is None:
    raise RuntimeError(
        "AdaptiveTolerance instance not found. "
        "Check that Cell 3 was executed successfully."
    )

learned_tolerance = float(adaptive_model.tolerance)

print("Learned tolerance:", learned_tolerance)

# ------------------------------------------------------------
# 2. Generate independent held-out residual data
# ------------------------------------------------------------

N = 10000

# Clean residuals: small numerical errors
clean = 10 ** rng.uniform(-16, -14, N)

# Corrupted residuals: substantially larger errors
corrupted = 10 ** rng.uniform(-8, -2, N)

residuals = np.concatenate([clean, corrupted])

# 1 = valid, 0 = invalid
labels = np.concatenate([
    np.ones(N, dtype=int),
    np.zeros(N, dtype=int)
])

# Shuffle independently
indices = rng.permutation(len(residuals))
residuals = residuals[indices]
labels = labels[indices]

# ------------------------------------------------------------
# 3. Prediction functions
# ------------------------------------------------------------

def adaptive_predict(residual):
    probability = float(
        adaptive_model.probability_valid(float(residual))
    )
    return int(probability >= 0.5)


def fixed_predict(residual, threshold):
    return int(residual <= threshold)


adaptive_predictions = np.array([
    adaptive_predict(r) for r in residuals
])

FIXED_THRESHOLD = 1e-11

fixed_predictions = np.array([
    fixed_predict(r, FIXED_THRESHOLD)
    for r in residuals
])

# ------------------------------------------------------------
# 4. Evaluation metrics
# ------------------------------------------------------------

def evaluate_predictions(name, predictions, actual):

    tp = int(np.sum((predictions == 1) & (actual == 1)))
    tn = int(np.sum((predictions == 0) & (actual == 0)))
    fp = int(np.sum((predictions == 1) & (actual == 0)))
    fn = int(np.sum((predictions == 0) & (actual == 1)))

    accuracy = (tp + tn) / len(actual) * 100

    precision = (
        tp / (tp + fp) * 100
        if tp + fp else 0
    )

    recall = (
        tp / (tp + fn) * 100
        if tp + fn else 0
    )

    print("\n", name)
    print("-" * 50)
    print("Accuracy :", f"{accuracy:.4f}%")
    print("Precision:", f"{precision:.4f}%")
    print("Recall   :", f"{recall:.4f}%")
    print("True valid:", tp)
    print("True invalid:", tn)
    print("False valid:", fp)
    print("False invalid:", fn)

    return accuracy


# ------------------------------------------------------------
# 5. Compare both systems
# ------------------------------------------------------------

adaptive_accuracy = evaluate_predictions(
    "ADAPTIVE TOLERANCE",
    adaptive_predictions,
    labels
)

fixed_accuracy = evaluate_predictions(
    "FIXED TOLERANCE",
    fixed_predictions,
    labels
)

difference = adaptive_accuracy - fixed_accuracy

# ------------------------------------------------------------
# 6. Final report
# ------------------------------------------------------------

print("\n" + "=" * 75)
print("COMPARISON SUMMARY")
print("=" * 75)

print("Adaptive threshold:", learned_tolerance)
print("Fixed threshold:", FIXED_THRESHOLD)

print("Adaptive accuracy:", f"{adaptive_accuracy:.4f}%")
print("Fixed accuracy:", f"{fixed_accuracy:.4f}%")

print("Accuracy difference:", f"{difference:+.4f} percentage points")

if adaptive_accuracy > fixed_accuracy:
    print("RESULT: Adaptive performed better on this test.")
elif adaptive_accuracy < fixed_accuracy:
    print("RESULT: Fixed tolerance performed better on this test.")
else:
    print("RESULT: Both performed equally on this test.")

print("=" * 75)

print(
    "NOTE: This is a synthetic residual-classification benchmark. "
    "It does not establish real-world solver validation improvement."
)
````

````python
# ============================================================
# SUPERNOVA TERAILLM PHOTON-1 V5
# HUGGING FACE AUTHENTICATION + PUBLICATION
# ============================================================

import os
from getpass import getpass
from huggingface_hub import HfApi

# Repository configuration
REPO_ID = "Supernova11c/Supernova-Photon-1-V5"
FOLDER = "/content/photon1_v5_hf"

print("=" * 65)
print("PHOTON-1 V5 — HUGGING FACE PUBLICATION")
print("=" * 65)

# Verify local files
required_files = [
    "README.md",
    "config.json",
    "requirements.txt",
    "photon1_v5.pt"
]

for filename in required_files:
    path = os.path.join(FOLDER, filename)

    if not os.path.isfile(path):
        raise FileNotFoundError(f"Missing file: {filename}")

    print("Verified:", filename)

# Secure token input
token = getpass("Enter your Hugging Face WRITE token: ")

if not token.strip():
    raise ValueError("Empty token")

# Authenticate
api = HfApi(token=token)

try:
    identity = api.whoami()
    username = identity["name"]

    print("\nAuthentication: SUCCESS")
    print("Account:", username)

except Exception as error:
    raise RuntimeError(
        "Authentication failed. Generate a valid WRITE token."
    ) from error

# Ensure repository ownership
if username.lower() != "supernova11c":
    raise PermissionError(
        "Authenticated account is not Supernova11c."
    )

# Create repository
api.create_repo(
    repo_id=REPO_ID,
    repo_type="model",
    private=False,
    exist_ok=True
)

print("\nRepository created/verified.")

# Upload all files
result = api.upload_folder(
    folder_path=FOLDER,
    repo_id=REPO_ID,
    repo_type="model",
    commit_message="Publish Supernova TeraiLLM Photon-1 V5"
)

print("\n" + "=" * 65)
print("PUBLICATION SUCCESS")
print("=" * 65)
print("Repository:", REPO_ID)
print("URL:", f"https://huggingface.co/{REPO_ID}")
print("Uploaded files:", len(required_files))
print("Commit:", result.oid)
print("=" * 65)
````
