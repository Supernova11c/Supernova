````python
# ============================================================
# CELL 2: PERSISTENT CHECKPOINT MANAGER
# ============================================================

class AstrovectorCheckpointManager:

    def __init__(self, home):

        self.home = Path(home)

        self.project_root = self.home / "projects"
        self.checkpoint_root = self.home / "checkpoints"

        self.project_root.mkdir(
            parents=True,
            exist_ok=True
        )

        self.checkpoint_root.mkdir(
            parents=True,
            exist_ok=True
        )

        self.excluded_directories = {
            ".git",
            "__pycache__",
            ".venv",
            "node_modules",
            ".next",
            ".cache"
        }

    def _safe_name(self, name):

        name = re.sub(
            r"[^a-zA-Z0-9_-]+",
            "-",
            str(name).strip()
        ).strip("-")

        if not name:
            raise ValueError("Invalid project name")

        return name[:80]

    def create_project(self, name):

        name = self._safe_name(name)
        project_path = self.project_root / name

        project_path.mkdir(
            parents=True,
            exist_ok=True
        )

        metadata = {
            "project_name": name,
            "created_at": datetime.now(
                timezone.utc
            ).isoformat(),
            "last_updated": datetime.now(
                timezone.utc
            ).isoformat(),
            "status": "CREATED"
        }

        metadata_path = project_path / ".astrovector.json"

        if not metadata_path.exists():
            metadata_path.write_text(
                json.dumps(metadata, indent=2),
                encoding="utf-8"
            )

        return project_path

    def _file_hash(self, path):

        digest = hashlib.sha256()

        with open(path, "rb") as file:

            for chunk in iter(
                lambda: file.read(1024 * 1024),
                b""
            ):
                digest.update(chunk)

        return digest.hexdigest()

    def create_checkpoint(self, project_name):

        project_name = self._safe_name(project_name)

        source = self.project_root / project_name

        if not source.is_dir():
            raise FileNotFoundError(source)

        checkpoint_id = (
            datetime.now(timezone.utc).strftime(
                "%Y%m%dT%H%M%S"
            )
            + "_"
            + uuid.uuid4().hex[:8]
        )

        temporary = self.checkpoint_root / (
            ".tmp_" + checkpoint_id
        )

        final = self.checkpoint_root / project_name / checkpoint_id

        temporary.mkdir(parents=True)

        snapshot = temporary / "project"

        def ignore(directory, names):

            return {
                name for name in names
                if name in self.excluded_directories
            }

        shutil.copytree(
            source,
            snapshot,
            ignore=ignore
        )

        manifest = []

        for file_path in snapshot.rglob("*"):

            if not file_path.is_file():
                continue

            manifest.append({
                "path": str(
                    file_path.relative_to(snapshot)
                ),
                "size": file_path.stat().st_size,
                "sha256": self._file_hash(file_path)
            })

        metadata = {
            "checkpoint_id": checkpoint_id,
            "project_name": project_name,
            "created_at": datetime.now(
                timezone.utc
            ).isoformat(),
            "file_count": len(manifest),
            "files": manifest
        }

        (temporary / "checkpoint.json").write_text(
            json.dumps(metadata, indent=2),
            encoding="utf-8"
        )

        final.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        temporary.rename(final)

        return {
            "checkpoint_id": checkpoint_id,
            "path": str(final),
            "file_count": len(manifest),
            "status": "SAVED"
        }

    def list_checkpoints(self, project_name):

        project_name = self._safe_name(project_name)

        directory = self.checkpoint_root / project_name

        if not directory.exists():
            return []

        checkpoints = []

        for item in sorted(directory.iterdir()):

            metadata_file = item / "checkpoint.json"

            if not metadata_file.is_file():
                continue

            metadata = json.loads(
                metadata_file.read_text(
                    encoding="utf-8"
                )
            )

            checkpoints.append(metadata)

        return checkpoints

    def restore_checkpoint(
        self,
        project_name,
        checkpoint_id,
        new_project_name=None
    ):

        project_name = self._safe_name(project_name)

        checkpoint_id = str(checkpoint_id)

        if not re.fullmatch(
            r"[A-Za-z0-9T_-]+",
            checkpoint_id
        ):
            raise ValueError("Invalid checkpoint ID")

        checkpoint = (
            self.checkpoint_root
            / project_name
            / checkpoint_id
        )

        metadata_path = checkpoint / "checkpoint.json"
        snapshot = checkpoint / "project"

        if not metadata_path.is_file() or not snapshot.is_dir():
            raise FileNotFoundError(
                "Checkpoint not found or incomplete"
            )

        metadata = json.loads(
            metadata_path.read_text(
                encoding="utf-8"
            )
        )

        # Verify saved file integrity before restoring.
        for entry in metadata["files"]:

            file_path = snapshot / entry["path"]

            if not file_path.is_file():
                raise ValueError(
                    "Checkpoint integrity failure: missing file"
                )

            if self._file_hash(file_path) != entry["sha256"]:
                raise ValueError(
                    "Checkpoint integrity failure: hash mismatch"
                )

        target_name = self._safe_name(
            new_project_name or project_name
        )

        target = self.project_root / target_name

        if target.exists():
            raise FileExistsError(
                "Restore target already exists: " + str(target)
            )

        shutil.copytree(
            snapshot,
            target
        )

        return {
            "project_name": target_name,
            "restored_from": checkpoint_id,
            "status": "RESTORED"
        }


# Initialize manager
astrovector_checkpoints = AstrovectorCheckpointManager(
    ASTROVECTOR_HOME
)

print("Checkpoint manager initialized.")
print("Persistent project storage:", astrovector_checkpoints.project_root)
print("Checkpoint storage:", astrovector_checkpoints.checkpoint_root)
````

````python
# ============================================================
# ASTROVECTOR 0.1
# CELL 3: PERSISTENT PROJECT AND TASK MEMORY
# ============================================================

class AstrovectorProjectMemory:

    def __init__(self, home):

        self.root = Path(home) / "memory" / "tasks"
        self.root.mkdir(parents=True, exist_ok=True)

    def _safe_id(self, value):

        value = str(value)

        if not re.fullmatch(r"[a-zA-Z0-9_-]{1,100}", value):
            raise ValueError("Invalid identifier")

        return value

    def _atomic_write(self, path, data):

        path.parent.mkdir(parents=True, exist_ok=True)

        temporary = path.with_name(
            path.name + ".tmp"
        )

        temporary.write_text(
            json.dumps(data, indent=2, ensure_ascii=False),
            encoding="utf-8"
        )

        temporary.replace(path)

    def create_task(self, project_name, objective):

        project_name = self._safe_id(project_name)

        if not isinstance(objective, str) or not objective.strip():
            raise ValueError("Task objective cannot be empty")

        task_id = uuid.uuid4().hex

        now = datetime.now(timezone.utc).isoformat()

        state = {
            "task_id": task_id,
            "project_name": project_name,
            "objective": objective.strip(),
            "status": "PLANNING",
            "plan": [],
            "completed_steps": [],
            "pending_steps": [],
            "last_checkpoint": None,
            "created_at": now,
            "updated_at": now
        }

        task_dir = self.root / project_name / task_id
        task_dir.mkdir(parents=True, exist_ok=False)

        self._atomic_write(
            task_dir / "state.json",
            state
        )

        self.record_event(
            project_name,
            task_id,
            "TASK_CREATED",
            {"objective": objective.strip()}
        )

        return state

    def _state_path(self, project_name, task_id):

        project_name = self._safe_id(project_name)
        task_id = self._safe_id(task_id)

        return (
            self.root
            / project_name
            / task_id
            / "state.json"
        )

    def load_task(self, project_name, task_id):

        path = self._state_path(project_name, task_id)

        if not path.is_file():
            raise FileNotFoundError("Task not found")

        return json.loads(
            path.read_text(encoding="utf-8")
        )

    def update_task(self, project_name, task_id, **changes):

        allowed = {
            "status",
            "plan",
            "completed_steps",
            "pending_steps",
            "last_checkpoint"
        }

        if not set(changes).issubset(allowed):
            raise ValueError("Unsupported state field")

        state = self.load_task(project_name, task_id)

        state.update(changes)

        state["updated_at"] = datetime.now(
            timezone.utc
        ).isoformat()

        self._atomic_write(
            self._state_path(project_name, task_id),
            state
        )

        return state

    def record_event(
        self,
        project_name,
        task_id,
        event_type,
        details=None
    ):

        project_name = self._safe_id(project_name)
        task_id = self._safe_id(task_id)

        task_dir = self.root / project_name / task_id
        task_dir.mkdir(parents=True, exist_ok=True)

        event = {
            "timestamp": datetime.now(
                timezone.utc
            ).isoformat(),
            "event": str(event_type),
            "details": details or {}
        }

        log_path = task_dir / "events.jsonl"

        with log_path.open("a", encoding="utf-8") as file:
            file.write(json.dumps(event, ensure_ascii=False) + "\n")
            file.flush()

        return event

    def list_tasks(self, project_name):

        project_name = self._safe_id(project_name)

        project_dir = self.root / project_name

        if not project_dir.exists():
            return []

        tasks = []

        for task_dir in project_dir.iterdir():

            state_path = task_dir / "state.json"

            if state_path.is_file():
                tasks.append(
                    json.loads(
                        state_path.read_text(encoding="utf-8")
                    )
                )

        return sorted(
            tasks,
            key=lambda item: item["created_at"],
            reverse=True
        )

    def get_events(self, project_name, task_id):

        project_name = self._safe_id(project_name)
        task_id = self._safe_id(task_id)

        log_path = (
            self.root
            / project_name
            / task_id
            / "events.jsonl"
        )

        if not log_path.exists():
            return []

        events = []

        for line in log_path.read_text(
            encoding="utf-8"
        ).splitlines():

            if line.strip():
                events.append(json.loads(line))

        return events


# Initialize persistent memory
astrovector_memory = AstrovectorProjectMemory(
    ASTROVECTOR_HOME
)

print("Astrovector persistent memory initialized.")
print("Storage:", astrovector_memory.root)
````

````python
# ============================================================
# ASTROVECTOR 0.1
# CELL 5: AUTONOMOUS DEVELOPMENT PLANNER
# ============================================================

class AstrovectorDevelopmentPlanner:

    CATEGORIES = {
        "research",
        "architecture",
        "environment",
        "implementation",
        "testing",
        "debugging",
        "documentation",
        "delivery"
    }

    def __init__(self, llm):

        self.llm = llm
        self.max_steps = 14

    def _extract_json(self, response):

        decoder = json.JSONDecoder()

        for index, character in enumerate(response):

            if character != "{":
                continue

            try:
                data, _ = decoder.raw_decode(
                    response[index:]
                )

                if isinstance(data, dict):
                    if "project_summary" in data and "steps" in data:
                        return data

            except json.JSONDecodeError:
                continue

        raise ValueError(
            "Planner did not return valid structured JSON."
        )

    def _validate_plan(self, plan):

        if not isinstance(plan, dict):
            raise ValueError("Invalid plan")

        summary = plan.get("project_summary")

        if not isinstance(summary, str) or not summary.strip():
            raise ValueError("Missing project summary")

        if len(summary) > 1500:
            raise ValueError("Project summary too long")

        assumptions = plan.get("assumptions", [])

        if not isinstance(assumptions, list) or len(assumptions) > 8:
            raise ValueError("Invalid assumptions")

        steps = plan.get("steps")

        if not isinstance(steps, list):
            raise ValueError("Steps must be a list")

        if not 3 <= len(steps) <= self.max_steps:
            raise ValueError("Invalid number of development steps")

        validated = []
        seen_ids = set()

        for index, step in enumerate(steps):

            if not isinstance(step, dict):
                raise ValueError("Invalid step structure")

            step_id = step.get("id")

            if not isinstance(step_id, str):
                raise ValueError("Invalid step ID")

            if not re.fullmatch(r"S[0-9]{1,2}", step_id):
                raise ValueError("Malformed step ID")

            if step_id in seen_ids:
                raise ValueError("Duplicate step ID")

            if step_id != f"S{index + 1}":
                raise ValueError("Step IDs must be sequential")

            title = step.get("title")
            description = step.get("description")
            category = step.get("category")

            if not isinstance(title, str) or not title.strip():
                raise ValueError("Missing step title")

            if not isinstance(description, str) or not description.strip():
                raise ValueError("Missing step description")

            if len(title) > 150 or len(description) > 1500:
                raise ValueError("Step content exceeds limits")

            if category not in self.CATEGORIES:
                raise ValueError(
                    f"Unsupported task category: {category}"
                )

            dependencies = step.get("depends_on", [])

            if not isinstance(dependencies, list):
                raise ValueError("Invalid dependencies")

            if len(set(dependencies)) != len(dependencies):
                raise ValueError("Duplicate dependencies")

            # A step can only depend on earlier steps.
            if not set(dependencies).issubset(seen_ids):
                raise ValueError(
                    f"Invalid dependency in {step_id}"
                )

            criteria = step.get("acceptance_criteria")

            if not isinstance(criteria, list):
                raise ValueError("Invalid acceptance criteria")

            if not 1 <= len(criteria) <= 5:
                raise ValueError("Each step needs 1-5 criteria")

            if not all(
                isinstance(item, str)
                and item.strip()
                and len(item) <= 500
                for item in criteria
            ):
                raise ValueError("Invalid acceptance criterion")

            validated.append({
                "id": step_id,
                "title": title.strip(),
                "description": description.strip(),
                "category": category,
                "depends_on": dependencies,
                "acceptance_criteria": criteria
            })

            seen_ids.add(step_id)

        return {
            "project_summary": summary.strip(),
            "assumptions": assumptions,
            "steps": validated
        }

    def generate_plan(self, objective):

        if not isinstance(objective, str) or not objective.strip():
            raise ValueError("Empty development request")

        if len(objective) > 4000:
            raise ValueError("Development request too long")

        prompt = f"""
You are the autonomous development planning engine
of Supernova Astrovector.

Your purpose is to plan professional full-stack
software and website development.

USER DEVELOPMENT REQUEST:
{objective}

Create a realistic, sequential development plan.

Requirements:

1. Understand the complete request.
2. Choose an appropriate architecture.
3. Include research when necessary.
4. Include environment preparation.
5. Include frontend and backend implementation
   when the request requires them.
6. Include database work when necessary.
7. Include testing and debugging.
8. Include documentation and delivery.
9. Avoid unnecessary steps.
10. Make each step independently understandable.

Return ONLY a JSON object.

Required structure:

{{
  "project_summary": "Project description",
  "assumptions": ["Assumption"],
  "steps": [
    {{
      "id": "S1",
      "title": "Step title",
      "description": "Detailed work description",
      "category": "architecture",
      "depends_on": [],
      "acceptance_criteria": [
        "Measurable completion condition"
      ]
    }}
  ]
}}

Allowed categories:
research, architecture, environment, implementation,
testing, debugging, documentation, delivery.

Generate between 3 and 14 steps.

IDs must be sequential: S1, S2, S3...

Dependencies must reference earlier step IDs only.

Do not claim that any work has already been executed.
"""

        response = self.llm(
            prompt,
            max_new_tokens=1200
        )

        raw_plan = self._extract_json(response)

        return self._validate_plan(raw_plan)


# Initialize with existing Qwen inference function
astrovector_planner = AstrovectorDevelopmentPlanner(
    llm=qwen_generate
)

print("Astrovector development planner initialized.")
print("Planning engine: Qwen3.5-4B")
print("Maximum plan steps:", astrovector_planner.max_steps)
````

````python
# ============================================================
# ASTROVECTOR 0.1
# CELL 6: AUTONOMOUS DEVELOPMENT CONTROLLER
# ============================================================

class AstrovectorAutonomousController:

    def __init__(
        self,
        planner,
        memory,
        checkpoint_manager
    ):

        self.planner = planner
        self.memory = memory
        self.checkpoints = checkpoint_manager

    def start_task(self, project_name, objective):

        project_path = self.checkpoints.create_project(
            project_name
        )

        task = self.memory.create_task(
            project_name,
            objective
        )

        self.memory.record_event(
            project_name,
            task["task_id"],
            "DEVELOPMENT_STARTED",
            {
                "project_path": str(project_path)
            }
        )

        return task

    def plan_task(self, project_name, task_id):

        task = self.memory.load_task(
            project_name,
            task_id
        )

        self.memory.update_task(
            project_name,
            task_id,
            status="PLANNING"
        )

        self.memory.record_event(
            project_name,
            task_id,
            "PLANNING_STARTED"
        )

        plan = self.planner.generate_plan(
            task["objective"]
        )

        # Persist complete plan independently.
        task_dir = (
            self.memory.root
            / project_name
            / task_id
        )

        plan_path = task_dir / "plan.json"

        temporary = task_dir / "plan.json.tmp"

        temporary.write_text(
            json.dumps(plan, indent=2, ensure_ascii=False),
            encoding="utf-8"
        )

        temporary.replace(plan_path)

        self.memory.update_task(
            project_name,
            task_id,
            status="PLANNED",
            plan=plan["steps"],
            completed_steps=[],
            pending_steps=plan["steps"]
        )

        self.memory.record_event(
            project_name,
            task_id,
            "PLAN_CREATED",
            {
                "step_count": len(plan["steps"]),
                "plan_path": str(plan_path)
            }
        )

        return plan

    def checkpoint_task(self, project_name, task_id):

        task = self.memory.load_task(
            project_name,
            task_id
        )

        result = self.checkpoints.create_checkpoint(
            project_name
        )

        self.memory.update_task(
            project_name,
            task_id,
            last_checkpoint=result["checkpoint_id"]
        )

        self.memory.record_event(
            project_name,
            task_id,
            "CHECKPOINT_CREATED",
            result
        )

        return result

    def recover_task(self, project_name, task_id):

        task = self.memory.load_task(
            project_name,
            task_id
        )

        plan_path = (
            self.memory.root
            / project_name
            / task_id
            / "plan.json"
        )

        if not plan_path.is_file():
            raise FileNotFoundError(
                "Saved development plan unavailable"
            )

        plan = json.loads(
            plan_path.read_text(encoding="utf-8")
        )

        return {
            "task": task,
            "plan": plan,
            "events": self.memory.get_events(
                project_name,
                task_id
            )
        }


# Initialize controller
astrovector_controller = AstrovectorAutonomousController(
    planner=astrovector_planner,
    memory=astrovector_memory,
    checkpoint_manager=astrovector_checkpoints
)

print("Astrovector autonomous controller initialized.")
print("Persistent planning: ENABLED")
print("Task recovery: ENABLED")
print("Checkpoint integration: ENABLED")
````

````python
# ============================================================
# SUPERNOVA AUTONOMOUS — ASTROVECTOR
# EXECUTION ENGINE V0.1
# ============================================================

from pathlib import Path
from datetime import datetime, timezone
import json
import re
import uuid


class AstrovectorExecutionEngine:

    def __init__(
        self,
        llm,
        memory,
        checkpoint_manager
    ):
        self.llm = llm
        self.memory = memory
        self.checkpoints = checkpoint_manager

        self.supported_categories = {
            "research",
            "architecture",
            "documentation"
        }

        self.artifact_directory = "artifacts"

    def _safe_filename(self, value):

        value = re.sub(
            r"[^a-zA-Z0-9_-]+",
            "_",
            value.lower()
        )

        return value[:80] or "artifact"

    def _get_workspace(self, project_name):

        project_path = self.checkpoints.create_project(
            project_name
        )

        workspace = (
            Path(project_path) / self.artifact_directory
        )

        workspace.mkdir(
            parents=True,
            exist_ok=True
        )

        return workspace

    def _generate_artifact(self, objective, step):

        prompt = f"""
You are Astrovector, an autonomous software development agent.

Your task is to execute ONE development-plan step by
producing a useful project artifact.

PROJECT OBJECTIVE:
{objective}

CURRENT STEP:
{json.dumps(step, ensure_ascii=False, indent=2)}

INSTRUCTIONS:
1. Perform the requested work in the form of a document.
2. Use specific, technically meaningful information.
3. Include relevant decisions and requirements.
4. Do not claim that software was installed, deployed,
   tested, or executed unless that actually happened.
5. Do not use placeholder sentences.
6. Return the artifact content only.
7. Use Markdown formatting where appropriate.

Produce a complete, useful artifact for this step.
"""

        result = self.llm(
            prompt,
            max_new_tokens=1400
        )

        if not isinstance(result, str):
            raise TypeError("Model returned a non-text response.")

        result = result.strip()

        if len(result) < 100:
            raise ValueError(
                "Generated artifact is too short."
            )

        return result

    def execute_next_step(
        self,
        project_name,
        task_id
    ):

        task = self.memory.load_task(
            project_name,
            task_id
        )

        if not task:
            raise ValueError("Task not found.")

        if task["status"] not in {
            "PLANNED",
            "IN_PROGRESS"
        }:
            raise ValueError(
                f"Task cannot execute from status: {task['status']}"
            )

        pending = task.get("pending_steps", [])

        if not pending:
            return {
                "status": "NO_PENDING_STEPS",
                "message": "All planned steps are complete."
            }

        step = pending[0]

        category = step.get("category")

        if category not in self.supported_categories:
            raise NotImplementedError(
                f"Execution adapter not yet available for: {category}"
            )

        step_id = step["id"]

        print("=" * 55)
        print("ASTROVECTOR EXECUTION ENGINE")
        print("=" * 55)
        print("Task:", task_id)
        print("Step:", step_id)
        print("Title:", step["title"])
        print("Category:", category)
        print("\nGenerating project artifact...")

        self.memory.record_event(
            project_name,
            task_id,
            "STEP_EXECUTION_STARTED",
            {
                "step_id": step_id,
                "category": category
            }
        )

        try:

            artifact = self._generate_artifact(
                task["objective"],
                step
            )

            filename = (
                self._safe_filename(step_id + "_" + step["title"])
                + ".md"
            )

            workspace = self._get_workspace(
                project_name
            )

            artifact_path = workspace / filename

            # Write atomically.
            temporary_path = artifact_path.with_suffix(
                ".tmp"
            )

            temporary_path.write_text(
                artifact,
                encoding="utf-8"
            )

            temporary_path.replace(artifact_path)

            # Update persistent task state only after
            # the artifact has been successfully written.
            completed = list(
                task.get("completed_steps", [])
            )

            completed.append(step_id)

            remaining = pending[1:]

            new_status = (
                "COMPLETED"
                if not remaining
                else "IN_PROGRESS"
            )

            self.memory.update_task(
                project_name,
                task_id,
                status=new_status,
                completed_steps=completed,
                pending_steps=remaining
            )

            self.memory.record_event(
                project_name,
                task_id,
                "STEP_EXECUTION_COMPLETED",
                {
                    "step_id": step_id,
                    "artifact": str(artifact_path),
                    "artifact_bytes": artifact_path.stat().st_size
                }
            )

            checkpoint_id = self.checkpoints.create_checkpoint(
                project_name
            )

            self.memory.update_task(
                project_name,
                task_id,
                last_checkpoint=checkpoint_id
            )

            print("\nSTEP EXECUTED SUCCESSFULLY")
            print("Artifact:", artifact_path)
            print("Size:", artifact_path.stat().st_size, "bytes")
            print("Checkpoint:", checkpoint_id)
            print("Remaining steps:", len(remaining))
            print("Task status:", new_status)

            return {
                "status": "SUCCESS",
                "step_id": step_id,
                "artifact_path": str(artifact_path),
                "checkpoint_id": checkpoint_id,
                "remaining_steps": len(remaining)
            }

        except Exception as error:

            self.memory.record_event(
                project_name,
                task_id,
                "STEP_EXECUTION_FAILED",
                {
                    "step_id": step_id,
                    "error": str(error)
                }
            )

            print("\nSTEP EXECUTION FAILED")
            print("Reason:", str(error))

            raise


# Initialize execution engine.

astrovector_executor = AstrovectorExecutionEngine(
    llm=qwen_generate,
    memory=astrovector_memory,
    checkpoint_manager=astrovector_checkpoints
)

print("ASTROVECTOR EXECUTION ENGINE INITIALIZED")
print("Model: Qwen3.5-4B")
print("Persistent memory: ENABLED")
print("Automatic checkpointing: ENABLED")
print("Artifact generation: ENABLED")
````

````python
# ============================================================
# ASTROVECTOR — QUALITY GATE V0.1
# ============================================================

import json
import re
from pathlib import Path


class AstrovectorQualityGate:

    def __init__(self, llm):
        self.llm = llm
        self.max_attempts = 2

    def _extract_json(self, response):

        decoder = json.JSONDecoder()

        for i, char in enumerate(response):
            if char == "{":
                try:
                    data, _ = decoder.raw_decode(response[i:])
                    if isinstance(data, dict):
                        return data
                except json.JSONDecodeError:
                    continue

        raise ValueError("Quality evaluator returned invalid JSON.")

    def evaluate(self, objective, step, artifact):

        criteria = step["acceptance_criteria"]

        prompt = f"""
You are Astrovector's independent quality evaluator.

Evaluate the artifact against the actual acceptance criteria.

PROJECT OBJECTIVE:
{objective}

STEP:
{step["title"]}

ACCEPTANCE CRITERIA:
{json.dumps(criteria, ensure_ascii=False)}

ARTIFACT:
{artifact[:14000]}

Return ONLY valid JSON in this exact format:
{{
  "passed": false,
  "results": [
    {{
      "criterion": "Criterion text",
      "passed": false,
      "evidence": "Specific evidence or missing information"
    }}
  ],
  "missing_items": ["Missing requirement"],
  "overall_feedback": "Short explanation"
}}

RULES:
- Evaluate every criterion separately.
- Do not assume something exists unless the artifact proves it.
- Be strict and objective.
- A criterion passes only when supported by clear evidence.
- Overall passed must be true ONLY if every criterion passes.
- Do not reward length or impressive wording.
"""

        response = self.llm(
            prompt,
            max_new_tokens=700
        )

        result = self._extract_json(response)

        if not isinstance(result.get("passed"), bool):
            raise ValueError("Evaluator did not return a boolean result.")

        results = result.get("results")

        if not isinstance(results, list) or len(results) != len(criteria):
            raise ValueError("Evaluator returned an incomplete criteria assessment.")

        for item in results:
            if not isinstance(item, dict):
                raise ValueError("Invalid criterion assessment.")

            if not isinstance(item.get("passed"), bool):
                raise ValueError("Invalid criterion result.")

        # Recalculate the overall decision ourselves.
        # The model cannot pass a step if any criterion failed.
        all_passed = all(
            item["passed"] for item in results
        )

        result["passed"] = (
            result["passed"] and all_passed
        )

        return result


astrovector_quality = AstrovectorQualityGate(
    llm=qwen_generate
)


# ------------------------------------------------------------
# RETROSPECTIVE AUDIT OF THE ALREADY COMPLETED S1 ARTIFACT
# ------------------------------------------------------------

audit_task = astrovector_memory.load_task(
    "astrovector_first_builder_test",
    "14030b69a67f42b2aa5aaad2b02d6548"
)

s1_step = next(
    step for step in audit_task["plan"]
    if step["id"] == "S1"
)

s1_path = Path(
    "/content/drive/MyDrive/Supernova-Astrovector/projects/"
    "astrovector_first_builder_test/artifacts/"
    "s1_requirements_analysis.md"
)

s1_content = s1_path.read_text(encoding="utf-8")

s1_audit = astrovector_quality.evaluate(
    audit_task["objective"],
    s1_step,
    s1_content
)

print("=" * 55)
print("ASTROVECTOR — S1 QUALITY AUDIT")
print("=" * 55)

for item in s1_audit["results"]:
    status = "PASS" if item["passed"] else "FAIL"
    print(f"\n[{status}] {item['criterion']}")
    print("Evidence:", item.get("evidence", ""))

print("\nOVERALL:", "PASS" if s1_audit["passed"] else "NEEDS IMPROVEMENT")
print("Feedback:", s1_audit.get("overall_feedback", ""))


# ------------------------------------------------------------
# QUALITY-GATED EXECUTION FOR FUTURE STEPS
# ------------------------------------------------------------

def astrovector_execute_verified(project_name, task_id):

    task = astrovector_memory.load_task(
        project_name,
        task_id
    )

    if task["status"] not in {"PLANNED", "IN_PROGRESS"}:
        raise ValueError("Task is not ready for execution.")

    pending = task.get("pending_steps", [])

    if not pending:
        return {"status": "NO_PENDING_STEPS"}

    step = pending[0]

    if step["category"] not in astrovector_executor.supported_categories:
        raise NotImplementedError(
            f"No execution adapter for {step['category']} yet."
        )

    workspace = astrovector_executor._get_workspace(project_name)

    filename = (
        astrovector_executor._safe_filename(
            step["id"] + "_" + step["title"]
        ) + ".md"
    )

    artifact_path = workspace / filename

    for attempt in range(1, astrovector_quality.max_attempts + 1):

        print(f"\nAttempt {attempt}/{astrovector_quality.max_attempts}")
        print("Executing:", step["title"])

        astrovector_memory.record_event(
            project_name,
            task_id,
            "VERIFIED_EXECUTION_STARTED",
            {"step_id": step["id"], "attempt": attempt}
        )

        artifact = astrovector_executor._generate_artifact(
            task["objective"],
            step
        )

        print("Evaluating acceptance criteria...")

        evaluation = astrovector_quality.evaluate(
            task["objective"],
            step,
            artifact
        )

        if evaluation["passed"]:

            temporary_path = artifact_path.with_suffix(".tmp")

            temporary_path.write_text(
                artifact,
                encoding="utf-8"
            )

            temporary_path.replace(artifact_path)

            completed = list(task.get("completed_steps", []))
            completed.append(step["id"])

            remaining = pending[1:]

            new_status = (
                "COMPLETED" if not remaining else "IN_PROGRESS"
            )

            astrovector_memory.update_task(
                project_name,
                task_id,
                status=new_status,
                completed_steps=completed,
                pending_steps=remaining
            )

            astrovector_memory.record_event(
                project_name,
                task_id,
                "VERIFIED_STEP_COMPLETED",
                {
                    "step_id": step["id"],
                    "attempt": attempt,
                    "artifact": str(artifact_path)
                }
            )

            checkpoint = astrovector_checkpoints.create_checkpoint(
                project_name
            )

            checkpoint_id = checkpoint["checkpoint_id"]

            astrovector_memory.update_task(
                project_name,
                task_id,
                last_checkpoint=checkpoint_id
            )

            print("\nQUALITY GATE: PASSED")
            print("Artifact:", artifact_path)
            print("Checkpoint:", checkpoint_id)
            print("Remaining steps:", len(remaining))

            return {
                "status": "VERIFIED_SUCCESS",
                "step_id": step["id"],
                "attempt": attempt,
                "evaluation": evaluation,
                "artifact_path": str(artifact_path),
                "checkpoint": checkpoint
            }

        print("\nQUALITY GATE: FAILED")
        print(evaluation.get("overall_feedback", ""))

        for missing in evaluation.get("missing_items", []):
            print("Missing:", missing)

        astrovector_memory.record_event(
            project_name,
            task_id,
            "QUALITY_GATE_FAILED",
            {
                "step_id": step["id"],
                "attempt": attempt,
                "evaluation": evaluation
            }
        )

    # Crucially, the step remains pending.
    raise RuntimeError(
        f"Step {step['id']} failed quality evaluation after "
        f"{astrovector_quality.max_attempts} attempts. "
        "Task memory was not marked complete."
    )


print("\nASTROVECTOR QUALITY GATE INITIALIZED.")
print("Existing S1 audit:", "PASSED" if s1_audit["passed"] else "NEEDS IMPROVEMENT")
print("Future execution: acceptance-gated")
````

````python
# ============================================================
# ASTROVECTOR — QUALITY EVALUATOR V0.2
# Strict acceptance-criteria isolation
# ============================================================

def _astrovector_evaluate_strict(self, objective, step, artifact):

    criteria = step["acceptance_criteria"]

    prompt = f"""
You are an independent acceptance-test evaluator.

Your job is to determine whether ONE specific development
step satisfies its explicitly listed acceptance criteria.

PROJECT OBJECTIVE (context only):
{objective}

CURRENT STEP:
{step["title"]}

DESCRIPTION:
{step["description"]}

THE ONLY ACCEPTANCE CRITERIA:
{json.dumps(criteria, ensure_ascii=False)}

ARTIFACT:
{artifact[:14000]}

CRITICAL EVALUATION RULES:

1. Evaluate ONLY the listed acceptance criteria.
2. The project objective is context, NOT an additional checklist.
3. Never invent additional requirements.
4. Never penalize an artifact for missing features assigned
   to other development steps.
5. If a criterion requests multiple items, verify each item.
6. Use evidence from the artifact.
7. Do not demand implementation details when the criterion
   only requests documentation.
8. Every criterion must pass for the overall result to pass.

Return ONLY valid JSON:

{{
  "passed": true,
  "results": [
    {{
      "criterion": "Exact acceptance criterion",
      "passed": true,
      "evidence": "Specific evidence from the artifact"
    }}
  ],
  "missing_items": [],
  "overall_feedback": "Short evaluation"
}}

Include exactly one result for each acceptance criterion.
"""

    response = self.llm(
        prompt,
        max_new_tokens=700
    )

    result = self._extract_json(response)

    if not isinstance(result.get("passed"), bool):
        raise ValueError("Invalid overall evaluation.")

    results = result.get("results")

    if not isinstance(results, list) or len(results) != len(criteria):
        raise ValueError("Incomplete criterion evaluation.")

    for item in results:
        if not isinstance(item, dict):
            raise ValueError("Invalid evaluation item.")

        if not isinstance(item.get("passed"), bool):
            raise ValueError("Invalid criterion status.")

    result["passed"] = (
        result["passed"]
        and all(item["passed"] for item in results)
    )

    return result


# Install corrected evaluator.
AstrovectorQualityGate.evaluate = _astrovector_evaluate_strict

# Re-audit the original S1 artifact.
s1_audit = astrovector_quality.evaluate(
    audit_task["objective"],
    s1_step,
    s1_content
)

print("=" * 55)
print("ASTROVECTOR — STRICT S1 RE-EVALUATION")
print("=" * 55)

for item in s1_audit["results"]:
    print(
        "\n[" + ("PASS" if item["passed"] else "FAIL") + "]",
        item["criterion"]
    )
    print("Evidence:", item["evidence"])

print("\nOVERALL:", "PASS" if s1_audit["passed"] else "NEEDS IMPROVEMENT")
print("Feedback:", s1_audit["overall_feedback"])
````

````python
# ============================================================
# ASTROVECTOR — ENVIRONMENT DISCOVERY ENGINE V0.1
# ============================================================

import os
import sys
import shutil
import subprocess
import json
import platform
import importlib.util
from pathlib import Path
from datetime import datetime, timezone


class AstrovectorEnvironmentDiscovery:

    TOOL_CHECKS = {
        "python": ["--version"],
        "node": ["--version"],
        "npm": ["--version"],
        "npx": ["--version"],
        "pnpm": ["--version"],
        "yarn": ["--version"],
        "bun": ["--version"],
        "git": ["--version"],
        "docker": ["--version"],
        "podman": ["--version"],
        "java": ["-version"],
        "javac": ["-version"],
        "go": ["version"],
        "rustc": ["--version"],
        "cargo": ["--version"],
        "php": ["--version"],
        "ruby": ["--version"],
        "bundle": ["--version"],
        "sqlite3": ["--version"],
        "psql": ["--version"],
        "gcc": ["--version"],
        "g++": ["--version"],
        "cmake": ["--version"],
        "make": ["--version"],
        "uv": ["--version"],
        "poetry": ["--version"],
        "conda": ["--version"]
    }

    def __init__(self):
        self.report = {}

    def _check_tool(self, name, args):

        executable = shutil.which(name)

        if not executable:
            return {
                "available": False,
                "path": None,
                "version": None
            }

        try:
            result = subprocess.run(
                [executable] + args,
                capture_output=True,
                text=True,
                timeout=8,
                check=False
            )

            output = (
                result.stdout.strip()
                or result.stderr.strip()
            )

            return {
                "available": result.returncode == 0,
                "path": executable,
                "version": output.splitlines()[0][:200]
                    if output else "Version unavailable"
            }

        except Exception as error:
            return {
                "available": True,
                "path": executable,
                "version": "Detection error: " + str(error)
            }

    def scan(self):

        print("Scanning available development tools...")

        tools = {}

        for name, args in self.TOOL_CHECKS.items():
            tools[name] = self._check_tool(name, args)

        self.report = {
            "timestamp": datetime.now(
                timezone.utc
            ).isoformat(),
            "platform": platform.platform(),
            "python_version": sys.version,
            "python_executable": sys.executable,
            "working_directory": os.getcwd(),
            "tools": tools
        }

        return self.report


astrovector_environment_discovery = (
    AstrovectorEnvironmentDiscovery()
)

environment_report = (
    astrovector_environment_discovery.scan()
)

print("\n========== ENVIRONMENT INVENTORY ==========")

for name, info in environment_report["tools"].items():

    if info["available"]:
        print(
            f"[AVAILABLE] {name}: {info['version']}"
        )

    else:
        print(f"[MISSING] {name}")

print("\nPlatform:", environment_report["platform"])
print("Python:", environment_report["python_version"].split()[0])
````

````python
# ============================================================
# ASTROVECTOR — ENVIRONMENT READINESS ASSESSMENT V0.1
# ============================================================

import json
import shutil
import subprocess
from pathlib import Path

PROJECT_NAME = "astrovector_first_builder_test"

project_path = Path(
    "/content/drive/MyDrive/Supernova-Astrovector/projects"
) / PROJECT_NAME

inventory_path = (
    project_path / "environment" / "toolchain_inventory.json"
)

if not inventory_path.exists():
    raise FileNotFoundError("Toolchain inventory missing.")

with open(inventory_path, "r", encoding="utf-8") as f:
    inventory = json.load(f)

tools = inventory["tools"]

required_tools = {
    "node": "Node.js runtime",
    "npm": "JavaScript package manager",
    "git": "Version control"
}

optional_tools = {
    "docker": "Container infrastructure",
    "pnpm": "Alternative package manager",
    "bun": "Alternative JavaScript runtime"
}

def inspect_tool(name):

    existing = tools.get(name, {})

    current_path = shutil.which(name)

    if not current_path:
        return {
            "available": False,
            "version": None,
            "path": None
        }

    try:
        result = subprocess.run(
            [current_path, "--version"],
            capture_output=True,
            text=True,
            timeout=10
        )

        version = (
            result.stdout.strip()
            or result.stderr.strip()
        )

        return {
            "available": result.returncode == 0,
            "version": version.splitlines()[0][:150],
            "path": current_path
        }

    except Exception as error:
        return {
            "available": False,
            "version": str(error),
            "path": current_path
        }


readiness = {
    "required": {},
    "optional": {},
    "ready": True
}

print("=" * 55)
print("ASTROVECTOR ENVIRONMENT READINESS")
print("=" * 55)

for name, description in required_tools.items():

    result = inspect_tool(name)
    readiness["required"][name] = result

    if result["available"]:
        print(f"[READY] {description}: {result['version']}")
    else:
        print(f"[MISSING] {description}")
        readiness["ready"] = False

print("\nOPTIONAL TOOLS")

for name, description in optional_tools.items():

    result = inspect_tool(name)
    readiness["optional"][name] = result

    status = "AVAILABLE" if result["available"] else "NOT INSTALLED"

    print(f"[{status}] {description}")

readiness_path = (
    project_path
    / "environment"
    / "readiness_report.json"
)

readiness_path.write_text(
    json.dumps(readiness, indent=2),
    encoding="utf-8"
)

print("\n" + "=" * 55)
print("ENVIRONMENT READY:", readiness["ready"])
print("Report:", readiness_path)
````

````python
from pathlib import Path
import subprocess
import urllib.request
import urllib.error
import time
import json
import os
import signal

PROJECT = Path(
    "/content/astrovector_runtime/projects/astrovector_first_builder_test"
)

LOG_DIR = PROJECT / "logs"
LOG_DIR.mkdir(parents=True, exist_ok=True)

backend_log = open(LOG_DIR / "backend_runtime.log", "w")
frontend_log = open(LOG_DIR / "frontend_runtime.log", "w")

processes = {}
results = {}

def start_service(name, command, log_file):
    process = subprocess.Popen(
        command,
        cwd=PROJECT,
        stdout=log_file,
        stderr=subprocess.STDOUT,
        start_new_session=True
    )
    processes[name] = process
    print(f"Started {name}, PID: {process.pid}")

def check_http(name, url, expected_status=200):
    deadline = time.time() + 30
    last_error = None

    while time.time() < deadline:
        try:
            with urllib.request.urlopen(url, timeout=3) as response:
                body = response.read().decode("utf-8")
                results[name] = {
                    "status": "PASS" if response.status == expected_status else "FAIL",
                    "http_status": response.status,
                    "response": body
                }
                return
        except Exception as error:
            last_error = str(error)
            time.sleep(1)

    results[name] = {
        "status": "FAIL",
        "error": last_error
    }

try:
    start_service(
        "backend",
        ["npm", "run", "dev", "--workspace=backend"],
        backend_log
    )

    start_service(
        "frontend",
        [
            "npm", "exec", "--workspace=frontend",
            "--", "vite",
            "--host", "127.0.0.1",
            "--port", "5173",
            "--strictPort"
        ],
        frontend_log
    )

    check_http(
        "backend_health",
        "http://127.0.0.1:3000/api/v1/health"
    )

    check_http(
        "frontend_http",
        "http://127.0.0.1:5173/"
    )

finally:
    # Stop both services, including their child processes.
    for name, process in processes.items():
        try:
            os.killpg(process.pid, signal.SIGTERM)
        except ProcessLookupError:
            pass

    for process in processes.values():
        try:
            process.wait(timeout=5)
        except subprocess.TimeoutExpired:
            try:
                os.killpg(process.pid, signal.SIGKILL)
            except ProcessLookupError:
                pass

    backend_log.close()
    frontend_log.close()

# Save the test report.
report_path = PROJECT / "environment" / "runtime_test_report.json"
report_path.parent.mkdir(parents=True, exist_ok=True)
report_path.write_text(json.dumps(results, indent=2))

print("\n========== RUNTIME TEST RESULTS ==========")
for name, result in results.items():
    print(f"\n{name}: {result['status']}")
    print(json.dumps(result, indent=2))

print("\nALL RUNTIME TESTS PASSED:",
      bool(results) and all(x["status"] == "PASS" for x in results.values()))

print("\nLogs:", LOG_DIR)
print("Report:", report_path)
````

````python
from pathlib import Path
import subprocess
import json
import re
import time

PROJECT = Path(
    "/content/astrovector_runtime/projects/astrovector_first_builder_test"
)

APP_FILE = PROJECT / "frontend" / "src" / "App.tsx"
ARTIFACTS = PROJECT / "artifacts"
ARTIFACTS.mkdir(parents=True, exist_ok=True)

# Preserve the original starter component.
if not APP_FILE.exists():
    raise FileNotFoundError(APP_FILE)

BACKUP_FILE = ARTIFACTS / "App.tsx.initial_backup"
if not BACKUP_FILE.exists():
    BACKUP_FILE.write_text(APP_FILE.read_text())

requirements = """
Build a professional project-management dashboard using React and TypeScript.

Requirements:
1. Modern, polished SaaS dashboard layout.
2. Responsive sidebar navigation.
3. Dashboard overview with project statistics.
4. Project cards with progress indicators.
5. Recent task list with status and priority.
6. Team activity section.
7. Search input and functional navigation selection.
8. Interactive task status filtering.
9. Use realistic demonstration data.
10. Attractive typography, spacing, cards and responsive layout.

Technical constraints:
- Generate one self-contained React TSX file.
- Use only React and standard browser functionality.
- Use inline CSS or a style element.
- Do not import additional libraries.
- Do not use external images, APIs, or network requests.
- Do not use eval, Function constructor, or dynamic code execution.
- Do not access environment variables, cookies, or filesystem.
- Use valid TypeScript and JSX.
- Export the main component as the default export.
- Make the UI functional, not just a static screenshot.
- Return ONLY the complete TSX source code.
"""

def clean_generated_code(code):
    code = code.strip()

    # Remove Markdown fences if Qwen includes them.
    code = re.sub(r"^```(?:tsx|typescript|jsx)?\s*", "", code)
    code = re.sub(r"\s*```$", "", code)

    return code.strip()

def validate_generated_code(code):
    if len(code) < 500:
        raise ValueError("Generated component is unexpectedly short.")

    if "export default" not in code:
        raise ValueError("Missing default React component export.")

    forbidden = [
        "eval(",
        "new Function",
        "child_process",
        "process.env",
        "document.cookie",
        "XMLHttpRequest",
        "WebSocket("
    ]

    for item in forbidden:
        if item in code:
            raise ValueError(f"Rejected generated source: {item}")

    return True

def build_frontend():
    return subprocess.run(
        ["npm", "run", "build", "--workspace=frontend"],
        cwd=PROJECT,
        capture_output=True,
        text=True,
        timeout=180
    )

print("ASTROVECTOR: Starting autonomous implementation")
print("MODEL: Existing Qwen3.5-4B")
print("TARGET:", APP_FILE)

prompt = f"""
You are the frontend implementation engine inside Astrovector.

Generate the requested application component.

{requirements}
"""

success = False
build_output = ""
generated_code = ""

for attempt in range(1, 4):
    print(f"\n========== GENERATION ATTEMPT {attempt}/3 ==========")

    if attempt == 1:
        response = qwen_generate(prompt, max_new_tokens=3000)
    else:
        repair_prompt = f"""
You are repairing a React TypeScript application.

The previous generated code failed compilation.

COMPILER OUTPUT:
{build_output[-7000:]}

PREVIOUS SOURCE:
{generated_code}

Fix the compilation errors while preserving the dashboard features.

Return ONLY the complete corrected TSX source code.
Do not use additional dependencies or external APIs.
"""
        response = qwen_generate(repair_prompt, max_new_tokens=3000)

    candidate = clean_generated_code(response)

    try:
        validate_generated_code(candidate)
    except Exception as error:
        build_output = str(error)
        print("SOURCE VALIDATION FAILED:", error)
        continue

    generated_code = candidate

    # Save every accepted candidate.
    candidate_file = ARTIFACTS / f"qwen_dashboard_attempt_{attempt}.tsx"
    candidate_file.write_text(generated_code)

    # Write Qwen-generated source into the application.
    APP_FILE.write_text(generated_code)

    print("Qwen source written:", APP_FILE)
    print("Source size:", len(generated_code), "characters")
    print("Running frontend production build...")

    result = build_frontend()
    build_output = result.stdout + "\n" + result.stderr

    print(build_output[-6000:])

    if result.returncode == 0:
        success = True
        print("\nAUTONOMOUS IMPLEMENTATION BUILD: PASS")
        break

    print("\nBuild failed. Sending compiler feedback to Qwen.")

report = {
    "component": "frontend/src/App.tsx",
    "model": "Qwen3.5-4B",
    "attempts": attempt,
    "implementation_status": "SUCCESS" if success else "FAILED",
    "source_characters": len(generated_code),
    "build_exit_status": 0 if success else 1,
    "timestamp": time.strftime("%Y-%m-%d %H:%M:%S")
}

report_file = ARTIFACTS / "qwen_frontend_implementation_report.json"
report_file.write_text(json.dumps(report, indent=2))

print("\n========== ASTROVECTOR REPORT ==========")
print(json.dumps(report, indent=2))
print("Report:", report_file)
````

````python
import os
import re
import json
import time
import socket
import subprocess
import urllib.request
import urllib.error
from pathlib import Path

# Create an isolated test project.
DEBUG_TEST_DIR = ASTROVECTOR_RUNTIME / "astrovector_micro_debug_test"
DEBUG_TEST_DIR.mkdir(parents=True, exist_ok=True)

SERVER_FILE = DEBUG_TEST_DIR / "server.js"
PORT = 18765

# Ask Qwen to generate a small backend with one deliberate bug.
generation_prompt = """
Write a small Node.js HTTP backend using ONLY built-in Node.js modules.

Requirements:
1. GET /health returns JSON {"status":"healthy"}.
2. GET /sum?a=7&b=5 returns JSON {"result":2}.
3. The sum endpoint must deliberately contain the incorrect subtraction operation.
4. Read the port from process.env.PORT, defaulting to 8765.
5. Listen on 127.0.0.1.
6. Handle invalid numeric parameters with HTTP 400.
7. Return 404 for unknown routes.

Return ONLY the complete JavaScript source code.
No Markdown fences. No explanation.
"""

def extract_js(text):
    match = re.search(
        r"```(?:javascript|js)?\s*(.*?)```",
        text,
        flags=re.DOTALL | re.IGNORECASE
    )
    return match.group(1).strip() if match else text.strip()

print("ASTROVECTOR: Asking Qwen to generate backend...")

generated_response = qwen_generate(
    generation_prompt,
    max_new_tokens=900
)

generated_code = extract_js(generated_response)

assert "createServer" in generated_code, "No HTTP server found in Qwen output."

SERVER_FILE.write_text(generated_code, encoding="utf-8")

print("Generated source:", len(generated_code), "characters")
print("Saved:", SERVER_FILE)
print("\nInitial syntax validation...")

syntax = subprocess.run(
    ["node", "--check", str(SERVER_FILE)],
    capture_output=True,
    text=True
)

print("Syntax:", "PASS" if syntax.returncode == 0 else "FAIL")
if syntax.stderr:
    print(syntax.stderr)

# Test and repair loop.
repair_history = []
MAX_REPAIRS = 2

def run_backend_tests():
    env = os.environ.copy()
    env["PORT"] = str(PORT)

    syntax_check = subprocess.run(
        ["node", "--check", str(SERVER_FILE)],
        capture_output=True,
        text=True
    )

    if syntax_check.returncode != 0:
        return False, "SYNTAX ERROR:\n" + syntax_check.stderr

    process = subprocess.Popen(
        ["node", str(SERVER_FILE)],
        cwd=str(DEBUG_TEST_DIR),
        env=env,
        stdout=subprocess.PIPE,
        stderr=subprocess.PIPE,
        text=True
    )

    errors = []

    try:
        # Wait for the server to become available.
        health = None

        for _ in range(30):
            if process.poll() is not None:
                break

            try:
                with urllib.request.urlopen(
                    f"http://127.0.0.1:{PORT}/health",
                    timeout=1
                ) as response:
                    health = json.loads(response.read().decode())
                break
            except Exception:
                time.sleep(0.3)

        if health is None:
            errors.append("Server failed to start or /health did not respond.")
        elif health.get("status") != "healthy":
            errors.append(f"Health test failed: {health}")

        # Test the sum endpoint.
        try:
            with urllib.request.urlopen(
                f"http://127.0.0.1:{PORT}/sum?a=7&b=5",
                timeout=3
            ) as response:
                result = json.loads(response.read().decode())

            if result.get("result") != 12:
                errors.append(
                    f"SUM TEST FAILED: expected 12, received {result}"
                )

        except Exception as exc:
            errors.append(f"SUM ENDPOINT ERROR: {exc}")

    finally:
        process.terminate()

        try:
            process.wait(timeout=3)
        except subprocess.TimeoutExpired:
            process.kill()
            process.wait()

        stdout, stderr = process.communicate()

    if stderr:
        errors.append("SERVER STDERR:\n" + stderr)

    return len(errors) == 0, "\n".join(errors)

# Let Astrovector handle failures using Qwen.
for attempt in range(MAX_REPAIRS + 1):
    print(f"\n========== TEST ATTEMPT {attempt + 1} ==========")

    passed, test_errors = run_backend_tests()

    if passed:
        print("ALL BACKEND TESTS PASSED.")
        break

    print("TEST FAILURE:")
    print(test_errors)

    if attempt == MAX_REPAIRS:
        print("Repair limit reached.")
        break

    current_code = SERVER_FILE.read_text(encoding="utf-8")

    repair_prompt = f"""
You are Astrovector's autonomous debugging engine.

The following Node.js backend failed real execution tests.

ACTUAL TEST ERRORS:
{test_errors}

CURRENT COMPLETE SOURCE:
{current_code}

Repair the backend so that:
- GET /health returns {{"status":"healthy"}}.
- GET /sum?a=7&b=5 returns {{"result":12}}.
- Invalid inputs return HTTP 400.
- Unknown routes return HTTP 404.
- It uses only built-in Node.js modules.
- It listens on process.env.PORT or 8765.

Return the COMPLETE corrected JavaScript file.
Do not explain anything.
Do not use Markdown fences.
"""

    print("\nASTROVECTOR: Sending failure to Qwen for repair...")

    repair_response = qwen_generate(
        repair_prompt,
        max_new_tokens=1100
    )

    repaired_code = extract_js(repair_response)

    if "createServer" not in repaired_code:
        print("Repair rejected: no recognizable HTTP server.")
        break

    # Astrovector applies Qwen's generated repair automatically.
    SERVER_FILE.write_text(repaired_code, encoding="utf-8")

    repair_history.append({
        "attempt": attempt + 1,
        "errors": test_errors,
        "repaired_characters": len(repaired_code)
    })

    print("Qwen repair applied automatically.")

print("\n========== FINAL REPORT ==========")

final_passed, final_errors = run_backend_tests()

report = {
    "model": "Qwen3.5-4B",
    "project": "Astrovector Micro Debug Test",
    "generation": "Qwen",
    "execution": "Node.js",
    "testing": "HTTP endpoint assertions",
    "repair_engine": "Qwen-generated full-file replacement",
    "repair_attempts": len(repair_history),
    "final_status": "PASS" if final_passed else "FAIL",
    "repair_history": repair_history,
    "final_errors": final_errors
}

(DEBUG_TEST_DIR / "debug_report.json").write_text(
    json.dumps(report, indent=2),
    encoding="utf-8"
)

print(json.dumps(report, indent=2))
````
